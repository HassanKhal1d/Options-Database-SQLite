#### Import Libraries

```python
import pandas as pd
```

#### Import Data

```python
# Import cleaned data that we created from Data preprocessing notebook
options_data_clean = pd.read_csv(r"C:\Users\Hassa\OneDrive - Loughborough University\current\Algorthmic Trading for Beginners\Data Management\datamodules\minute_options_data\options_data_raw_cleaned.csv",
                                index_col=None)

# Set the correct datetime index
options_data_clean.index = pd.to_datetime(options_data_clean["expire_date"])
options_data_clean.index.name = ""

options_data_clean = options_data_clean.drop(columns=["Unnamed: 0"])
```

```python
# Define target date and strike price for filtering
date = ['2023-10-02','2023-10-03']
strike = 4300

# Filter df to select rows where index matches given dates ans strike price (k)
options_data_clean[(options_data_clean.index.normalize().isin(date)) &
                   (options_data_clean.strike == strike)][[
                                                         'symbol','strike','expire_date'
]]
```

#### 1) Filter based on Liquidity measures to remove Illiquid Options

```python
# Filter out options where the call ("c_size") or put ("p_size") has a size of '0 x 0'
# This removes contracts with no market depth (no available bids or asks)
options_data_clean = options_data_clean[(options_data_clean['c_size'] != '0 x 0')
                                    & (options_data_clean['p_size'] != '0 x 0')]

# Display the filtered options data
options_data_clean.head()
```

#### 2) Filter based on strike distance relative to underlying price to focus on relevant contracts

```python
# Filter out options where the strike distance percentage is greater than 5%
# This helps remove strikes that are too far from the underlying price, focusing on relevant contracts
options_data_clean = options_data_clean[
    options_data_clean.strike_distance_pct <= 0.05]

# Display the filtered options data
options_data_clean.head()
```

#### 3) Filter out all contract values =< 0 

```python
options_data_clean = options_data_clean[(options_data_clean["c_last"] > 0.0)
                                    & (options_data_clean["p_last"] > 0.0)]
options_data_clean.head()
```

#### 4) Filter based IV values to weed out unreliable data

```python
# Remove rows where either the call option's implied volatility ("c_iv") or the put option's implied volatility ("p_iv") is missing (NaN)
# This ensures we only work with options that have valid implied volatility values
options_data_clean = options_data_clean.dropna(subset=["c_iv", "p_iv"])

# Display the filtered options data
options_data_clean.head()
```

#### 5) Filter based on bid and ask values of calls and puts (remove ones with 0 bids or asks)

```python
# Remove options where either the call or put bid/ask prices are zero
# This ensures that only actively traded options with valid market quotes are kept
options_data_clean = options_data_clean[(options_data_clean.c_bid != 0.0) 
                                    & (options_data_clean.p_bid != 0.0) 
                                    & (options_data_clean.c_ask != 0.0) 
                                    & (options_data_clean.p_ask != 0.0)]

# Display the filtered options data
options_data_clean.head()
```

```python
# Display a concise summary of the options_data_clean
options_data_clean.info()
```

#### Adding columns

```python
# Compute call and put midpoints
options_data_clean['c_mid'] = (options_data_clean['c_bid'] + options_data_clean['c_ask']) / 2
options_data_clean['p_mid'] = (options_data_clean['p_bid'] + options_data_clean['p_ask']) / 2
```

```python
# Compute call and put spreads
options_data_clean['c_spread'] = options_data_clean['c_ask'] - options_data_clean['c_bid']
options_data_clean['p_spread'] = options_data_clean['p_ask'] - options_data_clean['p_bid']
```

#### Separate Calls and Puts Data

```python
# Selecting common columns
common_cols = [ 'symbol',
               'underlying_last', 'expire_date',  'expiry_type', 'dte', 'strike', 
               'strike_distance', 'strike_distance_pct']

# Selecting call option columns (columns starting with 'c_')
call_cols = common_cols + [col for col in options_data_clean.columns if col.startswith('c_')]

# Selecting put option columns (columns starting with 'p_')
put_cols = common_cols + [col for col in options_data_clean.columns if col.startswith('p_')]

# Creating call and put DataFrames
call_options_data = options_data_clean[call_cols].copy()
put_options_data = options_data_clean[put_cols].copy()

call_options_data.head()
```
